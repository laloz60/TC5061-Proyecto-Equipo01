# Evaluación P3 — clustering de estaciones

Este notebook gráfica los resultados de P3: reparto de estaciones por clúster y mapa espacial.

## Antes de correrlo

Sus insumos no están en el repo (`data/processed/p3_predictions.parquet` y `models/kmeans_p3.pkl`); se generan desde la raíz del repo:

```bash
python -m mibici.dataset             # si todavía no tienes los datos
python -m mibici.modeling.train p3   # entrena K-Means y serializa el modelo
python -m mibici.modeling.predict p3 # escribe las predicciones de las estaciones
```

## Ambiente

Se corre con el `.venv` del proyecto. Si el kernel no existe, se crea desde la raíz del repo:

```bash
.venv/bin/python -m ipykernel install --user --name mibici --display-name "Python (.venv) mibici"
```

Se elige en *Kernel → Change kernel* (Jupyter) o en *Select Kernel* (VS Code). Las figuras se guardan con las rutas de `mibici.config`, así que el notebook corre desde cualquier carpeta del proyecto.


````python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from mibici.config import FIGURES_DIR, PROCESSED_DATA_DIR

# Cargar las predicciones que generó el script de inferencia
df_preds = pd.read_parquet(PROCESSED_DATA_DIR / "p3_predictions.parquet")
display(df_preds.head())
````

````python
plt.figure(figsize=(8, 5))
sns.countplot(data=df_preds, x="cluster_assigned", palette="viridis")
plt.title("Distribución de Estaciones de MiBici por Clúster (P3)")
plt.xlabel("ID de Clúster")
plt.ylabel("Número de Estaciones")
plt.savefig(FIGURES_DIR / "p3_distribucion_clusters.png", dpi=300, bbox_inches="tight")
plt.show()
````

````python
plt.figure(figsize=(9, 7))
scatter = plt.scatter(
    df_preds["longitud"], 
    df_preds["latitud"], 
    c=df_preds["cluster_assigned"], 
    cmap="viridis", 
    s=60, 
    alpha=0.8
)
plt.colorbar(scatter, label="Clúster Asignado")
plt.title("Mapa Espacial de Clústeres de Estaciones - MiBici P3")
plt.xlabel("Longitud")
plt.ylabel("Latitud")
plt.savefig(FIGURES_DIR / "p3_mapa_geografico.png", dpi=300, bbox_inches="tight")
plt.show()
````
