Script para generar metricas de clustering

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import silhouette_score
from mibici.config import PROCESSED_DATA_DIR, UNIFIED_TRIPS_PATH

# Cargar las predicciones que generó el script de inferencia
df_preds = pd.read_parquet(PROCESSED_DATA_DIR / "p3_predictions.parquet")
display(df_preds.head())
```

```python
plt.figure(figsize=(8, 5))
sns.countplot(data=df_preds, x="cluster_assigned", palette="viridis")
plt.title("Distribución de Estaciones de MiBici por Clúster (P3)")
plt.xlabel("ID de Clúster")
plt.ylabel("Número de Estaciones")
plt.savefig("../reports/figures/p3_distribucion_clusters.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
plt.figure(figsize=(9, 7))
scatter = plt.scatter(
    df_preds["longitud"], 
    df_preds["latitud"], 
    c=df_preds["cluster_assigned"], 
    cmap="viridis", 
    s=60, 
    alpha=0.8
)
plt.colorbar(scatter, label="Clúster Asignado")
plt.title("Mapa Espacial de Clústeres de Estaciones - MiBici P3")
plt.xlabel("Longitud")
plt.ylabel("Latitud")
plt.savefig("../reports/figures/p3_mapa_geografico.png", dpi=300, bbox_inches="tight")
plt.show()
```
